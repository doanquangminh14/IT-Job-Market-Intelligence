# 📊 Trực Quan Hóa & Theo Dõi Lịch Sử Chất Lượng Dữ Liệu (History Metrics)

Notebook này hỗ trợ theo dõi và trực quan hóa các chỉ số chất lượng dữ liệu cào từ file `reports/profiling/history_metrics.csv`:
- 📈 **Bảng tổng quan tương tác**: Xem các chỉ số mới nhất của TopCV, TopDev, VietnamWorks.
- 💰 **Phân tích cơ cấu lương**: So sánh tỷ lệ lương có số cụ thể vs lương "Thương lượng/Thoả thuận".
- 🌐 **Phân bố ngôn ngữ tuyển dụng**: Tỷ lệ Tiếng Việt (`vi`), Tiếng Anh (`en`), và Song ngữ (`mixed`).
- 🕒 **Theo dõi xu hướng (Trend Tracking)**: Nhận biết biến động chất lượng dữ liệu theo từng đợt cào.

```python
import os
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
from IPython.display import display, HTML

# 1. Xác định đường dẫn file history_metrics.csv
possible_paths = [
    os.path.abspath(os.path.join("..", "reports", "profiling", "history_metrics.csv")),
    os.path.abspath(os.path.join("reports", "profiling", "history_metrics.csv"))
]

csv_path = next((p for p in possible_paths if os.path.exists(p)), None)

if not csv_path:
    raise FileNotFoundError("Không tìm thấy file reports/profiling/history_metrics.csv!")

print(f"📂 Đang đọc dữ liệu từ: {csv_path}")
df = pd.read_csv(csv_path)
df['evaluated_at'] = pd.to_datetime(df['evaluated_at'])
print(f"✅ Đã tải thành công {len(df)} dòng dữ liệu lịch sử!")
```

## 1. Toàn Bộ Lịch Sử Các Phiên Kiểm Kê

```python
# Hiển thị toàn bộ bảng lịch sử đã sắp xếp theo thời gian mới nhất
df.sort_values(by="evaluated_at", ascending=False).reset_index(drop=True)
```

## 2. Bảng Chỉ Số Snapshot Phiên Mới Nhất

```python
# Lọc lấy phiên kiểm kê mới nhất
latest_time = df['evaluated_at'].max()
df_latest = df[df['evaluated_at'] == latest_time].copy()

print(f"🕒 Thời điểm cập nhật mới nhất: {latest_time}")

summary_cols = [
    "source", "total_listings", "total_details", "success_details", 
    "duplicate_urls", "has_numeric_salary_pct", "negotiable_salary_pct", 
    "valid_deadline_pct", "lang_vi_pct", "lang_en_pct", "lang_mixed_pct"
]

# Format hiển thị chuyên nghiệp với màu sắc trực quan
df_latest[summary_cols].style \
    .format({
        "total_listings": "{:,.0f}",
        "total_details": "{:,.0f}",
        "success_details": "{:,.0f}",
        "duplicate_urls": "{:,.0f}",
        "has_numeric_salary_pct": "{:.1f}%",
        "negotiable_salary_pct": "{:.1f}%",
        "valid_deadline_pct": "{:.1f}%",
        "lang_vi_pct": "{:.1f}%",
        "lang_en_pct": "{:.1f}%",
        "lang_mixed_pct": "{:.1f}%",
    }) \
    .background_gradient(subset=["has_numeric_salary_pct"], cmap="Blues") \
    .background_gradient(subset=["valid_deadline_pct"], cmap="Greens")
```

## 3. Biểu Đồ Trực Quan Hóa Chất Lượng Dữ Liệu

```python
# A. So sánh quy mô tin cào giữa 3 sàn
fig_volume = px.bar(
    df_latest,
    x="source",
    y=["total_listings", "success_details"],
    barmode="group",
    title="<b>Quy Mô Thu Thập: Số Listing và Detail Thành Công Từng Sàn</b>",
    labels={"value": "Số lượng tin", "source": "Sàn tuyển dụng", "variable": "Chỉ số"},
    color_discrete_map={"total_listings": "#3b82f6", "success_details": "#10b981"},
    text_auto=True
)
fig_volume.update_layout(template="plotly_white", legend_title_text="Chỉ số")
fig_volume.show()
```

```python
# B. Cơ cấu mức lương (Khoảng lương cụ thể vs Thương lượng)
fig_salary = px.bar(
    df_latest,
    x="source",
    y=["has_numeric_salary_pct", "negotiable_salary_pct"],
    barmode="stack",
    title="<b>Cơ Cấu Mức Lương: Có Số Cụ Thể vs Thương Lượng (%)</b>",
    labels={"value": "Tỷ lệ (%)", "source": "Sàn tuyển dụng", "variable": "Phân loại lương"},
    color_discrete_map={"has_numeric_salary_pct": "#0ea5e9", "negotiable_salary_pct": "#94a3b8"},
    text_auto=".1f"
)
fig_salary.update_layout(template="plotly_white", yaxis_range=[0, 105], legend_title_text="Loại lương")
fig_salary.show()
```

```python
# C. Phân bố ngôn ngữ tuyển dụng (Tiếng Việt vs Tiếng Anh vs Song ngữ)
fig_lang = px.bar(
    df_latest,
    x="source",
    y=["lang_vi_pct", "lang_en_pct", "lang_mixed_pct"],
    barmode="group",
    title="<b>Phân Bố Ngôn Ngữ Tuyển Dụng Theo Từng Sàn (%)</b>",
    labels={"value": "Tỷ lệ (%)", "source": "Sàn tuyển dụng", "variable": "Ngôn ngữ"},
    color_discrete_map={
        "lang_vi_pct": "#f59e0b", 
        "lang_en_pct": "#6366f1", 
        "lang_mixed_pct": "#ec4899"
    },
    text_auto=".1f"
)
fig_lang.update_layout(template="plotly_white", legend_title_text="Ngôn ngữ")
fig_lang.show()
```

## 4. Theo Dõi Xu Hướng Biến Động Theo Thời Gian (Trend Tracking)

```python
# D. Đường biểu diễn xu hướng chỉ số qua các mốc thời gian
fig_trend = px.line(
    df,
    x="evaluated_at",
    y="has_numeric_salary_pct",
    color="source",
    markers=True,
    title="<b>Biến Động Tỷ Lệ Tin Có Lương Cụ Thể Theo Thời Gian (%)</b>",
    labels={
        "has_numeric_salary_pct": "Tỷ lệ có lương (%)", 
        "evaluated_at": "Thời điểm kiểm kê", 
        "source": "Nguồn"
    }
)
fig_trend.update_layout(template="plotly_white")
fig_trend.show()
```
